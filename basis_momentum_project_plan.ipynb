## 1. Project objective and research claim

**Paper's claim:** Basis-momentum—the difference between trailing 12-month returns on first- and second-nearby commodity futures strategies—predicts commodity spot and term premiums beyond conventional basis and momentum characteristics, both over time and across commodities.

**Our goal:** Build a reproducible commodity futures-curve pipeline, replicate the assigned exhibits, update them through the latest available data, validate the source data independently, and deliver an **instructor-approved extension** that adds something beyond the paper.

**Scope:** Start the historical replication in **1986** (rather than the paper's July 1959 start) because of source coverage. Explain sample and commodity-universe differences in the report. The assigned exhibit numbers refer to the **2017 working paper**; the accepted manuscript renumbers exhibits. Verify all definitions against the correct version before coding.

## 4. Data sources, coverage, and construction

| Source | Intended role | Important caveat |
|---|---|---|
| **WRDS Datastream** (`tr_ds_fut`, `wrds_fut_contract`) | Historical futures continuation and contract-level data, targeting 1986 onward | Availability differs by commodity and exchange; verify actual dates and fields |
| **Databento** (`GLBX.MDP3`) | CME Group outright futures contracts from June 2010 onward; modern update and source reconciliation | Check which contracts/commodities are covered and which settlement or comparable price fields are available |
| **Bloomberg** (trading lab, if needed) | Historical COMEX metal coverage before 2010 | Contingency, not a guaranteed required feed |
| **Original paper sources** | Commodity Research Bureau and Futures Industry Institute | Not available through the course; numerical discrepancies must be documented |

**Required shared panel:** one row per `commodity × month × nearby_rank`, with at least `commodity`, `date`, `contract_id`, `nearby_rank` (1–4), `expiration`, `price`, `monthly_excess_return`, and `source`. A commodity-month characteristic table should contain `basis`, `momentum`, and `basis_momentum`.

**Implementation rules:**
- Build first through fourth nearby contracts using the paper's rule: hold the n-th nearby and roll out when it expires in month **t+2**.
- Compute each month's futures return on the **same contract** from one observation date to the next; do not accidentally treat the jump between two different contract prices as an investment return.
- Compute basis, 12-month momentum, and 12-month basis-momentum according to the paper's definitions, with lagged predictors for next-month tests.
- Establish the commodity universe, observation date, price conventions, missing-data handling, and any contract-specific exceptions **before** analyzing results.
